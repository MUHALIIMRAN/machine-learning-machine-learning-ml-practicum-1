```python
# Membuat matrix 5x5 dengan angka random 0-100
matrix = np.random.randint(0, 101, (5, 5))

print("Matrix 5x5:")
print(matrix)

# Transpose matrix
transpose = matrix.T

print("\nTranspose Matrix:")
print(transpose)

# Determinant matrix
determinant = np.linalg.det(matrix)

print("\nDeterminant Matrix:")
print(determinant)
```

# Tugas 4: Eksplorasi Dataset Wine

Pada tugas ini digunakan dataset Wine dari Scikit-learn untuk melihat data, statistik, dan karakteristik dataset.

```python
from sklearn.datasets import load_wine

# Load dataset Wine
wine = load_wine()

# Mengubah menjadi DataFrame
wine_df = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

# Menambahkan target
wine_df['target'] = wine.target

print("10 Baris Pertama Dataset Wine:")
display(wine_df.head(10))
```

```python
print("Statistik Deskriptif Dataset Wine:")
display(wine_df.describe())
```

```python
plt.figure(figsize=(8, 5))

wine_df['target'].value_counts().sort_index().plot(kind='bar')

plt.title('Distribusi Kelas Dataset Wine')
plt.xlabel('Kelas Wine')
plt.ylabel('Jumlah Data')
plt.xticks(
    ticks=[0, 1, 2],
    labels=['Kelas 0', 'Kelas 1', 'Kelas 2'],
    rotation=0
)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    wine_df['alcohol'],
    wine_df['malic_acid'],
    c=wine_df['target'],
    alpha=0.7
)

plt.title('Hubungan Alcohol dan Malic Acid')
plt.xlabel('Alcohol')
plt.ylabel('Malic Acid')
plt.colorbar(label='Kelas Wine')

plt.show()
```
